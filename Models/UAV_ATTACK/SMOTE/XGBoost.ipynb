```python
from sklearn.metrics import classification_report, accuracy_score, f1_score
```

```python
from gps_ids import protocol, training

EXP = protocol.smote_heldout("uav")
r = training.run("XGBoost (SMOTE)", EXP)
model = r.model
y_pred = r.y_pred
y_test = EXP.y_test
feature_names = EXP.feature_names
```

```python
# Evaluate the model

print("\nEvaluating the model on the test set...")
accuracy = accuracy_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred, average='weighted', zero_division=0)
print(f"Model Accuracy on Test Set: {accuracy * 100:.2f}%")
print(f"F1 Score: {f1:.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
# Seed sweep: the remaining seeds in training.SEEDS, appended to the results CSV.
SWEEP_SEEDS = [s for s in training.SEEDS if s != protocol.SEED]

training.run_seeds("XGBoost (SMOTE)", lambda s: protocol.smote_heldout("uav", seed=s), seeds=SWEEP_SEEDS)
```
