```python
from sklearn.metrics import classification_report, accuracy_score
```

```python
from gps_ids import protocol, training

EXP = protocol.ensemble_heldout("uav")
r = training.run("StackedEnsemble_RF (SMOTE)", EXP)
stacked_pred = r.y_pred
y_meta_test = EXP.y_test
```

```python
# Evaluation
report = classification_report(
    y_meta_test,
    stacked_pred,
    labels=[0, 1],  # 0 = benign, 1 = malicious
    target_names=["benign", "malicious"],
    zero_division=0,
    output_dict=True
)
f1 = report["weighted avg"]["f1-score"]
accuracy = accuracy_score(y_meta_test, stacked_pred)
print(classification_report(
    y_meta_test,
    stacked_pred,
    labels=[0, 1],
    target_names=["benign", "malicious"],
    zero_division=0
))
print(f"F1 Score: {f1:.4f}")
print(f"Accuracy: {accuracy:.4f}")
```

```python
# Seed sweep: the remaining seeds in training.SEEDS, appended to the results CSV.
SWEEP_SEEDS = [s for s in training.SEEDS if s != protocol.SEED]

training.run_seeds("StackedEnsemble_RF (SMOTE)", lambda s: protocol.ensemble_heldout("uav", seed=s), seeds=SWEEP_SEEDS)
```
