```python
from gps_ids import protocol, training
```

```python
feature_cols_1 = ['vdop', 'ax', 'hdop', 'cog_rad', 'jamming_indicator', 'terrain_alt_valid', 'dist_bottom_valid', 'terrain_alt', 'lat_x', 'lat_y']
feature_cols_2 = ['lat_y', 'vel_e_m_s', 'vel_n_m_s', 'c_variance_rad', 'hdop', 's_variance_m_s', 'q[0]', 'vel_m_s' ,'cog_rad', 'q[3]']

EXP1 = protocol.real_heldout("uav", features=feature_cols_1)
EXP2 = protocol.real_heldout("uav", features=feature_cols_2)
```

```python
for base in ("XGBoost", "MLP Classifier", "Logistic Regression", "Random Forest"):
    training.run(f"{base} (Explainability Features)", EXP1)
    training.run(f"{base} (Xgboost Features)", EXP2)
```

```python
# Seed sweep: the remaining seeds in training.SEEDS, appended to the results CSV.
SWEEP_SEEDS = [s for s in training.SEEDS if s != protocol.SEED]

for base in ("XGBoost", "MLP Classifier", "Logistic Regression", "Random Forest"):
    training.run_seeds(
        f"{base} (Explainability Features)",
        lambda s: protocol.real_heldout("uav", features=feature_cols_1, seed=s),
        seeds=SWEEP_SEEDS,
    )
    training.run_seeds(
        f"{base} (Xgboost Features)",
        lambda s: protocol.real_heldout("uav", features=feature_cols_2, seed=s),
        seeds=SWEEP_SEEDS,
    )
```
